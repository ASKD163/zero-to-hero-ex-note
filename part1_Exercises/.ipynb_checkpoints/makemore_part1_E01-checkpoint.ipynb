E01: train a trigram language model, i.e. take two characters as an input to predict the 3rd one. Feel free to use either counting or a neural net. Evaluate the loss; Did it improve over a bigram model?

**trigram**

```python
# open file
words = open('names.txt', 'r').read().splitlines()
words[:3]
```

```python
chars = sorted(list(set(''.join(words)))) # 样本里的所有字符
chars.insert(0, '.') # 让 sti 从 .. 开始
sti = {} # 前两个字符对应index
its = {} # index 到前两个字符
k = 0
for ch1 in chars:
    for ch2 in chars:
        firstTwo = ch1 + ch2
        sti[firstTwo] = k
        its[k] = firstTwo
        k += 1

# 单字符对应index
cti = {s:i for i,s in enumerate(chars)}
itc = {i:s for i,s in enumerate(chars)}
```

```python
xs, ys = [], []
for w in words:
    chs = ['.'] + ['.'] + list(w) + ['.']
    for i in range(0, len(chs) - 2):
        tmp = chs[i:i+3]
        firstTwo = tmp[0] + tmp[1]
        ch = tmp[2]
        xs.append(sti[firstTwo])
        ys.append(cti[ch])
        # print(f'{firstTwo}, {ch}')
xs = torch.tensor(xs)
ys = torch.tensor(ys)

num = xs.nelement()
print('number of examples: ', num)
```

```python
xenc = F.one_hot(xs, num_classes = 729).float()
```

```python
# Init Network
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((729, 27), generator = g, requires_grad = True)
```

```python
for i in range(20):
    # forward (softmax)
    logits = xenc @ W # log-counts
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdims = True)
    
    #loss
    loss = -probs[torch.arange(num), ys].log().mean()
    print(f'loss: {loss}')
    
    W.grad = None
    loss.backward()
    W.data += -80 * W.grad
print(f'loss: {loss}')
```

```python
# sample
outs = []
for i in range(0,3):
    out = '..'
    ix = 0
    while True:
        outxenc = F.one_hot(torch.tensor([ix]), num_classes = 729).float()
        logits = outxenc @ W
        counts = logits.exp()
        probs = counts / counts.sum(1, keepdims = True)
        i = torch.multinomial(probs, num_samples = 1, replacement = True, generator = g).item()
        out += itc[i]
        ix = sti[out[-2:]]
        # print(out[-2:])
        if (i == 0):
            break
    outs.append(out)
outs
```
